# Task 2 — Logical Operations: AND, OR, NOT



```python
# Step 1 — Import and Load Image(s)

import cv2
import numpy as np
import matplotlib.pyplot as plt

def show(images, titles, cols=4):
    n = len(images)
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(4 * cols, 4 * rows))
    for i, (img, title) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if len(img.shape) == 3:
            plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(img, cmap='gray', vmin=0, vmax=255)
        plt.title(title)
        plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
# Load both images and convert to grayscale
c1 = cv2.imread('image1.jpg')
c2 = cv2.imread('image2.jpg')
img1 = cv2.cvtColor(c1, cv2.COLOR_BGR2GRAY)
img2 = cv2.cvtColor(c2, cv2.COLOR_BGR2GRAY)
# Resize image2 to match image1
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

show([c1, img1, c2, img2], ['Image 1 (color)', 'Image 1 (gray)',
                            'Image 2 (color)', 'Image 2 (gray)'], cols=2)
```

## Step 2 — Convert to Binary

First we threshold both images to black & white (0 or 255), because logical operations work on binary values.

```python
_, b1 = cv2.threshold(img1, 127, 255, cv2.THRESH_BINARY)
_, b2 = cv2.threshold(img2, 127, 255, cv2.THRESH_BINARY)

show([b1, b2], ['Binary 1', 'Binary 2'], cols=2)
```

## Step 3 — Apply AND, OR, NOT

- **AND** — keeps white pixels present in BOTH images
- **OR**  — keeps white pixels present in EITHER image
- **NOT** — inverts the image (white↔black)

```python
and_img = cv2.bitwise_and(b1, b2)
or_img  = cv2.bitwise_or(b1, b2)
not_img = cv2.bitwise_not(b1)

show([and_img, or_img, not_img], ['AND', 'OR', 'NOT (of image 1)'])
```
