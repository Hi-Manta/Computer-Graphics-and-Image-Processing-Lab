# Experiment-5: Reflection_about_y-axis

```python
import cv2  #OpenCV library for python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/Kawasaki-H2-Carbon-New.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#Reflection about X-axis

reflected = cv2.flip(image, 1)
```

```python
plt.figure(figsize= (10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(reflected)
plt.title("Reflected Image")
plt.axis("off")

plt.show()
```
