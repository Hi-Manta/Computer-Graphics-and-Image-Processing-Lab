```python
import cv2  #OpenCV library for python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/ninja_h2r.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimension

h, w = image.shape[:2]
```

```python
#translation values

tx = 100
ty = 50
```

```python
#Tranlation matrix

M = np.float32(
    [[1, 0, tx], [0, 1, ty]]
)
```

```python
#Apply translation

translated = cv2.warpAffine(image, M, (w, h))
```

```python
#Display

plt.figure(figsize = (10, 5))
```

```python
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")

plt.show()
```
