```python
import cv2  #OpenCV library for python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/Kawasaki-H2-Carbon-New.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimension

h, w = image.shape[:2]
```

# Step-1: Scaling

```python
sx = 10
sy = 7


scaled = cv2.resize(image, (0,0), fx=sx, fy=sy)
```

# Step-2: Rotation

```python
h2, w2 = scaled.shape[:2]
```

```python
center = (w2//2, h2//2)
```

```python
M_rotation = cv2.getRotationMatrix2D(center, 45, 1)
```

```python
rotated = cv2.warpAffine(scaled, M_rotation, (w2, h2))
```

# Step-3: Translation

```python
tx = 50
ty = 30
```

```python
M_translation = np.float32([[1, 0, tx], [0, 1, ty]])
```

```python
transformed = cv2.warpAffine(rotated, M_translation, (w2+tx, h2+ty))
```

# Step-4: Reflection

```python
#Reflection about X-axis

reflected_x = cv2.flip(image, 0)
reflected_y = cv2.flip(image, 1)
```

```python
#Display All
plt.figure(figsize= (50, 50))

plt.subplot(1, 6, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 6, 4)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")

plt.subplot(1, 6, 3)
plt.imshow(rotated)
plt.title("Rotated Image")
plt.axis("off")

plt.subplot(1, 6, 2)
plt.imshow(transformed)
plt.title("Transformed Image")
plt.axis("off")

plt.subplot(1, 6, 5)
plt.imshow(reflected_x)
plt.title("Reflected Image X")
plt.axis("off")

plt.subplot(1, 6, 6)
plt.imshow(reflected_y)
plt.title("Reflected Image Y")
plt.axis("off")

plt.show()
```
