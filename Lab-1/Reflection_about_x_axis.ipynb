```python
import cv2  #OpenCV library for python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/ninja_h2r.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#Reflection about X-axis

reflected_x = cv2.flip(image, 0)
reflected_y = cv2.flip(image, 1)
```

```python
plt.figure(figsize= (25, 20))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(reflected_x)
plt.title("Reflected Image X")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(reflected_y)
plt.title("Reflected Image Y")
plt.axis("off")

plt.show()
```
