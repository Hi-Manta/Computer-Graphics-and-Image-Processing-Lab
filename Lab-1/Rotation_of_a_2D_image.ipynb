```python
import cv2  #OpenCV library for python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/ninja_h2r.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#image dimension

h, w = image.shape[:2]
```

```python
#center of image

center = (w// 2, h//2)
```

```python
#Rotation angle
angle = 90
```

```python
#Rotation matrix

M = cv2.getRotationMatrix2D(center, angle, 1.0)
```

```python
rotated = cv2.warpAffine(image, M, (w, h))
```

```python
plt.figure(figsize= (10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rotated)
plt.title("Rotated Image")
plt.axis("off")

plt.show()
```
